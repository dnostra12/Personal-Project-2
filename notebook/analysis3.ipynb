### 데이터 확인

2023년 데이터에서 동일 법인이 반복된 사례를 확인한 결과, 서로 다른 사업장이 실제로 존재하는 경우와 거의 동일한 값이 중복된 경우가 함께 나타났다.

기업 단위 분석에서는 `업체` 행이 있는 경우 업체 값을 우선 사용하고, 업체 행 없이 서로 다른 사업장이 명확한 경우에는 사업장 값을 합산하기로 했다.

반면 동일 사업장으로 보이지만 배출량이 거의 같은 값으로 반복된 일부 기업은 어떤 값이 최종값인지 확인하기 어려워 2023년 기업 단위 분석에서 제외하기로 했다.

```python
# 배출량과 에너지 사용량을 숫자로 변환
numeric_cols = [
    '온실가스 배출량(tCO₂eq)',
    '에너지 사용량(TJ)'
]

gir_2023_temp = gir_2023.copy()

for col in numeric_cols:
    gir_2023_temp[col] = pd.to_numeric(
        gir_2023_temp[col]
        .astype(str)
        .str.replace(',', '', regex=False),
        errors='coerce'
    )


# 번호만 다른 완전 중복 제거
check_cols = [
    col for col in gir_2023_temp.columns
    if col != '번호'
]

gir_2023_temp = gir_2023_temp.drop_duplicates(
    subset=check_cols,
    keep='first'
).copy()


# 값이 거의 같지만 어느 행이 최종값인지 판단하기 어려운 기업은 제외
ambiguous_2023 = [
    '에스케이피유코어 주식회사',
    '에이치디씨폴리올(주)',
    '쿠팡로지스틱스서비스 유한회사'
]

gir_2023_temp = gir_2023_temp[
    ~gir_2023_temp['법인명'].isin(ambiguous_2023)
].copy()
```

```python
company_rows = []

for company, group in gir_2023_temp.groupby('법인명'):

    # 같은 내용인데 비고만 다른 행은 한 번만 사용
    core_cols = [
        col for col in group.columns
        if col not in ['번호', '비고']
    ]

    group = group.drop_duplicates(
        subset=core_cols,
        keep='first'
    )

    # 업체 행이 있는 경우
    company_level = group[
        group['지정구분'] == '업체'
    ]

    if len(company_level) > 0:

        # 기업 단위 분석에서는 업체 값을 우선 사용
        row = company_level.iloc[0].copy()
        row['집계방식'] = '업체 우선'

    else:

        # 업체 행이 없으면 사업장 데이터 사용
        sites = group[
            group['지정구분'] == '사업장'
        ]

        if len(sites) == 1:

            row = sites.iloc[0].copy()
            row['집계방식'] = '단일 사업장'

        else:

            # 여러 사업장이 실제로 존재하면 합산
            row = sites.iloc[0].copy()

            row['온실가스 배출량(tCO₂eq)'] = (
                sites['온실가스 배출량(tCO₂eq)'].sum()
            )

            row['에너지 사용량(TJ)'] = (
                sites['에너지 사용량(TJ)'].sum()
            )

            row['집계방식'] = f'사업장 {len(sites)}개 합산'

    company_rows.append(row)


# 기업 단위 데이터 생성
gir_2023_company = pd.DataFrame(company_rows)

gir_2023_company.shape
```

```python
# GIR 데이터를 기업 단위로 정리하는 함수
def make_company_data(df):

    df = df.copy()

    # 숫자형 컬럼 정리
    numeric_cols = [
        '온실가스 배출량(tCO₂eq)',
        '에너지 사용량(TJ)'
    ]

    for col in numeric_cols:
        df[col] = pd.to_numeric(
            df[col]
            .astype(str)
            .str.replace(',', '', regex=False),
            errors='coerce'
        )

    # 번호와 비고만 다른 동일 데이터 제거
    check_cols = [
        col for col in df.columns
        if col not in ['번호', '비고']
    ]

    df = df.drop_duplicates(
        subset=check_cols,
        keep='first'
    ).copy()

    company_rows = []

    for company, group in df.groupby('법인명'):

        company_level = group[
            group['지정구분'] == '업체'
        ]

        # 업체 행이 1개이면 그대로 사용
        if len(company_level) == 1:

            row = company_level.iloc[0].copy()
            row['집계방식'] = '업체 우선'

        # 업체 행이 여러 개인 경우
        elif len(company_level) > 1:

            ghg_values = company_level[
                '온실가스 배출량(tCO₂eq)'
            ].dropna().unique()

            # 배출량 값이 모두 같으면 한 번만 사용
            if len(ghg_values) == 1:

                row = company_level.iloc[0].copy()

                row['온실가스 배출량(tCO₂eq)'] = ghg_values[0]

                # 에너지 값이 서로 다르면 임의 선택하지 않음
                energy_values = company_level[
                    '에너지 사용량(TJ)'
                ].dropna().unique()

                if len(energy_values) > 1:
                    row['에너지 사용량(TJ)'] = pd.NA

                row['집계방식'] = '업체 중복값 확인'

            else:
                # 서로 다른 업체 값이면 판단하지 않고 제외
                continue

        else:

            sites = group[
                group['지정구분'] == '사업장'
            ]

            if len(sites) == 1:

                row = sites.iloc[0].copy()
                row['집계방식'] = '단일 사업장'

            else:

                # 사업장 여러 개면 일단 법인 단위로 합산
                row = sites.iloc[0].copy()

                row['온실가스 배출량(tCO₂eq)'] = (
                    sites['온실가스 배출량(tCO₂eq)'].sum()
                )

                row['에너지 사용량(TJ)'] = (
                    sites['에너지 사용량(TJ)'].sum()
                )

                row['집계방식'] = f'사업장 {len(sites)}개 합산'

        company_rows.append(row)

    return pd.DataFrame(company_rows)
```

```python
gir_2022 = pd.read_csv("../data/gir_2022.csv")

gir_2022_company = make_company_data(gir_2022)

print('기업 단위 행 수:', len(gir_2022_company))
print('고유 법인 수:', gir_2022_company['법인명'].nunique())

print('\n집계방식')
print(gir_2022_company['집계방식'].value_counts())
```

## 2020~2025년 기업 단위 GIR 데이터 통합

연도별 GIR 데이터를 기업 단위로 정리한 뒤 하나의 데이터로 통합했다.
2023년에 중복 여부를 명확하게 판단하기 어려웠던 일부 기업은 잘못된 배출량 합산을 방지하기 위해 제외했다.

```python
# 2023년에 판단이 어려웠던 기업
ambiguous_2023 = [
    '에스케이피유코어 주식회사',
    '에이치디씨폴리올(주)',
    '쿠팡로지스틱스서비스 유한회사'
]

gir_company_list = []

for year in range(2020, 2026):

    # 연도별 데이터 불러오기
    df = pd.read_csv(
        f"../data/gir_{year}.csv"
    )

    # 2023년의 판단이 어려운 3개 기업만 제외
    if year == 2023:
        df = df[
            ~df['법인명'].isin(ambiguous_2023)
        ].copy()

    # 기업 단위 데이터로 정리
    company_df = make_company_data(df)

    gir_company_list.append(company_df)

    print(
        year,
        '기업 수:',
        len(company_df)
    )
```

```python
# 2020~2025년 기업 단위 데이터 통합
gir_company = pd.concat(
    gir_company_list,
    ignore_index=True
)

print('전체 행 수:', len(gir_company))

# 연도별 기업 수 확인
gir_company[
    '대상년도'
].value_counts().sort_index()
```

```python
# 한 기업이 같은 연도에 두 번 남아 있는지 확인
duplicate_company_year = gir_company.duplicated(
    subset=['법인명', '대상년도']
).sum()

print(
    '기업명 + 대상년도 중복:',
    duplicate_company_year
)
```

## 온실가스 배출량 연도별 현황

기업 단위로 정리한 GIR 데이터를 활용해 2020~2025년 온실가스 배출량의 기본 분포를 확인했다.

연도마다 관리 대상 기업의 구성이 다를 수 있기 때문에 총배출량뿐만 아니라 기업 수와 중앙값도 함께 확인했다.

```python
# 연도별 온실가스 배출량 기본 통계
ghg_year_summary = (
    gir_company
    .groupby('대상년도')
    .agg(
        기업수=('법인명', 'nunique'),
        총배출량=('온실가스 배출량(tCO₂eq)', 'sum'),
        평균배출량=('온실가스 배출량(tCO₂eq)', 'mean'),
        중앙값배출량=('온실가스 배출량(tCO₂eq)', 'median')
    )
    .round(1)
)

ghg_year_summary
```

### 분석 결과

2020~2025년 GIR 데이터를 확인한 결과, 분석 대상 기업 수는 1,060개에서 1,172개로 증가했다.

총 온실가스 배출량은 2021년에 가장 높게 나타난 이후 지속적으로 감소했고, 기업당 평균 배출량도 같은 기간 감소하는 흐름을 보였다.

반면 중앙값 배출량은 2023년까지 증가한 뒤 2024년과 2025년에 다시 감소했다.

다만 연도마다 GIR에 포함된 기업 구성이 다르기 때문에, 이러한 변화만으로 동일 기업들의 온실가스 배출량이 감소했다고 판단하기는 어렵다. 따라서 다음 분석에서는 여러 연도에 공통으로 포함된 기업만 따로 비교하기로 했다.

```python
# 기업별로 GIR 데이터에 등장한 연도 수 확인
company_year_count = (
    gir_company
    .groupby('법인명')['대상년도']
    .nunique()
)

# 2020~2025년 6개 연도 모두 존재하는 기업
common_companies = company_year_count[
    company_year_count == 6
].index

print('6개 연도 모두 존재하는 기업 수:', len(common_companies))

# 동일 기업 패널 데이터 생성
gir_balanced = gir_company[
    gir_company['법인명'].isin(common_companies)
].copy()

# 연도별 기업 수 확인
gir_balanced.groupby('대상년도')['법인명'].nunique()
```

## 동일 기업의 온실가스 배출량 변화

연도별 대상 기업 구성 차이의 영향을 줄이기 위해, 2020~2025년 6개 연도에 모두 포함된 기업만 따로 추출해 온실가스 배출량 변화를 비교했다.

```python
# 6개 연도 모두 포함된 동일 기업들의 배출량 요약
balanced_summary = (
    gir_balanced
    .groupby('대상년도')
    .agg(
        기업수=('법인명', 'nunique'),
        총배출량=('온실가스 배출량(tCO₂eq)', 'sum'),
        평균배출량=('온실가스 배출량(tCO₂eq)', 'mean'),
        중앙값배출량=('온실가스 배출량(tCO₂eq)', 'median')
    )
    .round(1)
)

balanced_summary
```

```python
# 2020년 대비 2025년 총배출량 변화율
ghg_2020 = balanced_summary.loc[2020, '총배출량']
ghg_2025 = balanced_summary.loc[2025, '총배출량']

change_rate = (
    (ghg_2025 - ghg_2020) / ghg_2020 * 100
)

print(
    '2020년 대비 2025년 총배출량 변화율:',
    round(change_rate, 1),
    '%'
)
```

### 분석 결과

2020~2025년 모든 연도에 공통으로 포함된 동일 기업만 비교한 결과, 2025년 총 온실가스 배출량은 2020년보다 9.4% 감소했다.

연도별 대상 기업 구성이 달라지는 영향을 줄인 뒤에도 배출량 감소가 나타났기 때문에, 전체 GIR 데이터에서 확인된 감소가 단순한 기업 구성 변화만으로 발생한 것은 아니었다.

다만 절대 배출량은 기업의 생산량이나 매출 규모 변화에도 영향을 받을 수 있기 때문에, 이를 곧바로 환경효율 개선으로 해석하지는 않았다.

```python
# 기업별 2020년과 2025년 배출량 비교
company_change = (
    gir_balanced[
        gir_balanced['대상년도'].isin([2020, 2025])
    ]
    .pivot(
        index='법인명',
        columns='대상년도',
        values='온실가스 배출량(tCO₂eq)'
    )
    .dropna()
)

# 배출량 변화 계산
company_change['변화량'] = (
    company_change[2025] - company_change[2020]
)

# 증가 / 감소 / 동일 구분
company_change['변화방향'] = company_change['변화량'].apply(
    lambda x: '감소' if x < 0
    else '증가' if x > 0
    else '동일'
)

# 기업 수와 비율 확인
change_direction = pd.DataFrame({
    '기업수': company_change['변화방향'].value_counts(),
    '비율(%)': (
        company_change['변화방향']
        .value_counts(normalize=True)
        .mul(100)
        .round(1)
    )
})

change_direction
```

### 분석 결과

2020년과 2025년의 동일 기업을 비교한 결과, 배출량이 감소한 기업은 51.2%, 증가한 기업은 48.6%로 기업 수 기준에서는 거의 비슷하게 나타났다.

따라서 전체 배출량이 9.4% 감소한 결과가 대부분의 기업에서 공통적으로 나타난 현상이라고 보기는 어렵다. 감소 기업의 감축 규모와 증가 기업의 증가 규모 차이가 전체 배출량 변화에 영향을 줬을 가능성이 있다.

```python
# 감소 기업의 총 감소량
total_decrease = (
    company_change.loc[
        company_change['변화량'] < 0,
        '변화량'
    ].sum()
)

# 증가 기업의 총 증가량
total_increase = (
    company_change.loc[
        company_change['변화량'] > 0,
        '변화량'
    ].sum()
)

# 전체 순변화량
net_change = company_change['변화량'].sum()

print('감소 기업 총 변화량:', round(total_decrease))
print('증가 기업 총 변화량:', round(total_increase))
print('전체 순변화량:', round(net_change))
```

### 분석 결과

배출량이 감소한 기업들의 총 감소량은 약 7,130만 tCO₂eq였고, 증가한 기업들의 총 증가량은 약 2,052만 tCO₂eq로 나타났다.

기업 수는 감소 기업과 증가 기업이 거의 비슷했지만, 감소 기업의 감축 규모가 더 크게 나타나면서 전체적으로 약 5,079만 tCO₂eq의 순감소가 발생했다.

따라서 전체 배출량 감소는 기업 수의 차이보다는 감소 기업과 증가 기업의 배출량 변화 규모 차이에서 나타난 것으로 볼 수 있다.


```python
# 배출량 감소폭이 큰 기업부터 정렬
top_reducers = (
    company_change
    .sort_values('변화량')
    .head(10)
    .copy()
)

top_reducers
```

```python
# 상위 10개 감소 기업의 총 감소량
top10_decrease = top_reducers['변화량'].sum()

# 전체 감소량 중 상위 10개 기업 비중
top10_share = (
    abs(top10_decrease) / abs(total_decrease) * 100
)

print(
    '상위 10개 감소 기업의 전체 감소량 비중:',
    round(top10_share, 1),
    '%'
)
```

### 분석 결과

배출량 감소폭이 큰 상위 10개 기업이 전체 감소량의 69.8%를 차지했다.

따라서 2020~2025년 동일 기업의 총배출량이 9.4% 감소했지만, 이러한 감소가 모든 기업에서 고르게 나타난 것은 아니었다. 전체 감소량의 상당 부분이 일부 대형 배출기업의 감축에서 발생한 것으로 나타났다.

기업 수 기준으로는 감소 기업과 증가 기업이 거의 비슷했기 때문에, 전체 배출량 감소를 국내 기업 전반의 공통적인 감축 흐름으로 확대해서 해석하지는 않았다.

```python
# 배출량 감소폭이 큰 상위 10개 기업 확인
top_reducers_table = (
    company_change
    .sort_values('변화량')
    .head(10)
    .reset_index()
)

# 기업별 전체 감소량 기여도 계산
top_reducers_table['감소기여도(%)'] = (
    abs(top_reducers_table['변화량'])
    / abs(total_decrease)
    * 100
).round(1)

top_reducers_table[
    [
        '법인명',
        2020,
        2025,
        '변화량',
        '감소기여도(%)'
    ]
]
```

```python
# 상위 10개 감소 기업의 총 감소량
top10_decrease = top_reducers_table['변화량'].sum()

# 전체 감소량 중 상위 10개 기업 비중
top10_share = (
    abs(top10_decrease)
    / abs(total_decrease)
    * 100
)

print(
    '상위 10개 감소 기업의 전체 감소량 비중:',
    round(top10_share, 1),
    '%'
)
```

```python
# 상위 10개 감소 기업의 총 감소량
top10_decrease = top_reducers_table['변화량'].sum()

# 전체 감소량 중 상위 10개 기업 비중
top10_share = (
    abs(top10_decrease)
    / abs(total_decrease)
    * 100
)

print(
    '상위 10개 감소 기업의 전체 감소량 비중:',
    round(top10_share, 1),
    '%'
)
```

```python
# 상위 10개 감소 기업 이름
top10_names = top_reducers_table['법인명']

# 상위 10개 기업 제외
remaining_change = company_change[
    ~company_change.index.isin(top10_names)
].copy()

# 나머지 기업의 순변화량
remaining_net_change = remaining_change['변화량'].sum()

# 나머지 기업의 2020년 총배출량
remaining_2020 = remaining_change[2020].sum()

# 변화율
remaining_change_rate = (
    remaining_net_change
    / remaining_2020
    * 100
)

print('상위 10개 제외 기업 수:', len(remaining_change))
print('상위 10개 제외 순변화량:', round(remaining_net_change))
print('상위 10개 제외 변화율:', round(remaining_change_rate, 1), '%')
```

```python
# 상위 10개 감소 기업 이름
top10_names = top_reducers_table['법인명']

# 상위 10개 기업 제외
remaining_change = company_change[
    ~company_change.index.isin(top10_names)
].copy()

# 나머지 기업의 순변화량
remaining_net_change = remaining_change['변화량'].sum()

# 나머지 기업의 2020년 총배출량
remaining_2020 = remaining_change[2020].sum()

# 2020년 대비 2025년 변화율
remaining_change_rate = (
    remaining_net_change
    / remaining_2020
    * 100
)

print('상위 10개 제외 기업 수:', len(remaining_change))
print('상위 10개 제외 순변화량:', round(remaining_net_change))
print('상위 10개 제외 변화율:', round(remaining_change_rate, 1), '%')
```
